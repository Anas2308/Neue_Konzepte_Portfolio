```python
import re
```

```python
# re.search(): Suche nach einem Wort im Text
# Die Funktion re.search gibt ein re.Match Objekt zurück mit Informationen zu Suche und Resultat

text = "Hallo, ich bin ein einfacher Regex-Test!"
muster = r"einfacher"
print(re.search(muster, text))

print('Gefundener String: ', re.search(muster, text).group())
print('Start- und Endposition des Matches: ', re.search(muster, text).span())
```

```python
# re.findall(): findet alle Vorkommen eines Musters
text = "Apfel, Banane, Apfel, Kirsche"
muster = r"Apfel"

print(re.findall(muster, text))
print(re.findall('.', text)) # beliebiges Zeichen 
print(re.findall('\w', text)) # beliebige alphanumerische Zeichen
print(re.findall('[abcde]', text)) # Zeichen abcde
print(re.findall('^Apfel', text)) # Apfel nur am Anfang
print(re.findall('^Banane', text)) # Banane nur am Anfang
print(re.findall('[^Banane]', text)) # Alle Zeichen außer Banane
```

```python
# Identifkation von 'the'
text = "The sun was shining brightly as they walked along the beach. She picked up a seashell from the sand and held it in her hand. There were a few clouds in the sky, but the weather remained warm. The sound of the waves crashing against the shore was soothing. In the distance, they could see a small boat sailing across the horizon."
muster = r"the"
print(re.findall(muster, text)) # Am Satzanfang wird the nicht gefunden, da groß geschriben, the aus weather wird ebefalls gefunden

muster = r"[Tt]he" # Groß- und Kleinschreibung berücksichtigen
print(re.findall(muster, text))

muster = r"\b[Tt]he\b" # Wortanfang und -ende mit \b und T oder t
print(re.findall(muster, text))
```

## Substitution

```python
# re.sub(): Ersetze ein Wort durch ein anderes
text = "Ich mag Katzen. Katzen sind süß."
muster = r"Katzen"

neuer_text = re.sub(muster, "Hunde", text)
print("Ersetzter Text:", neuer_text)
```

```python
# sub mit Capture Groups

text = "Ich mag Katzen. Katzen sind süß."
muster = r"(Katzen)" # Capture Group \1

neuer_text = re.sub(muster, "Hunde", text)
print("Ersetzter Text:", neuer_text)
```

```python
text = "the bigger they were, the smaller they will be"
muster = r"the (.*)er they were, the \1 they will be"

neuer_text = re.sub(muster, muster, text)
print(neuer_text)
```

```python
# Beispiele: ELIZA

text = "Men are all alike."
muster = r".* all .*"

neuer_text = re.sub(muster, r"IN WHAT WAY", text)

print("User: ", text)
print("ELIZA: ", neuer_text)
print("--------------------------------------")


text2 = "They’re always bugging us about something or other."
muster2 = r".* always .*"

neuer_text2 = re.sub(muster2, r"CAN YOU THINK OF A SPECIFIC EXAMPLE", text2)

print("User: ", text2)
print("ELIZA: ", neuer_text2)
print("--------------------------------------")

text3 = "Well, my boyfriend made me come here."
muster3a = r"\bmy\b"
muster3b = r"\bme\b"
muster3c = r"\b[Ww]ell\b"

neuer_text3 = re.sub(muster3a, r"YOUR", text3)
neuer_text3b = re.sub(muster3b, r"YOU", neuer_text3)
neuer_text3c = re.sub(r",", r"", re.sub(muster3c, r"", neuer_text3b)).upper() # entfernen des well,

# .upper() - umwandlung in Großbuchstaben

print("User: ", text3)
print("ELIZA: ", neuer_text3c)
print("--------------------------------------")


text4 = "He says I’m depressed much of the time"
muster4 = r".* (depressed|sad) .*"

# Capture Group \1 (das Muster) wird in den neuen Text eingefügt
neuer_text4 = re.sub(muster4, r"I AM SORRY TO HEAR YOU ARE \1", text4).upper()

print("User: ", text4)
print("ELIZA: ", neuer_text4)
print("--------------------------------------")
```

```python
# Capture Groups
# Umwandeln von einem Datum im Format DD-MM-YYYY in YYYY-MM-DD
date_text = "Today's date is 18-03-2025 and yesterday was 17-03-2025."

# Caputre Groups in runden Klammern
date_pattern = r"(\d{2})-(\d{2})-(\d{4})" # Caputre Group \1, \2, \3

reformatted_text = re.sub(date_pattern, r"\3-\2-\1", date_text) # Ersetzen des Textes mit Caputre Groups in umgekehrter Reihenfolge

print(reformatted_text)

#https://www.geeksforgeeks.org/python-regex-replace-captured-groups/
```
